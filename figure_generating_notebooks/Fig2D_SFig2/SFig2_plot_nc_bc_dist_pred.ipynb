```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['font.family'] = 'Arial'
mpl.rcParams['font.size'] = 7
mpl.rcParams['axes.labelsize'] = 7
mpl.rcParams['axes.titlesize'] = 7
mpl.rcParams['xtick.labelsize'] = 7
mpl.rcParams['ytick.labelsize'] = 7
mpl.rcParams['legend.fontsize'] = 7
mpl.rcParams['mathtext.fontset'] = 'custom'
mpl.rcParams['mathtext.rm'] = 'Arial'
mpl.rcParams['mathtext.it'] = 'Arial:italic'
mpl.rcParams['mathtext.bf'] = 'Arial:bold'
```

```python
nc_bc_mod = pd.read_csv("nc_barcoded_detection_obs_pred_mod.csv")
```

```python
nc_bc_mod
```

```python
fig_width_in = 170 / 25.4  # 170 mm
plt.figure(figsize=(fig_width_in, fig_width_in * 0.75))
plt.plot(nc_bc_mod['nc_counts'], nc_bc_mod['predicted'], '.', markersize=1, label=r"Negative control codeword (model $R^2 =0.51$)")
plt.legend()
plt.xlabel("Observed # of decoded barcodes")
plt.ylabel("Predicted # of decoded barcodes")
plt.xticks([100, 300, 500, 700, 900, 1100, 1300], minor=True)
plt.yticks([100, 300, 500, 700, 900, 1100, 1300], minor=True)
plt.savefig("SFIG2_nc_bc_dist_mod.pdf")
```
